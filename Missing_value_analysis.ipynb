```python
import pandas as pd

# Titanic dataset load
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# 1. Dataset overview
print("Shape:", df.shape)
df.info()

# 2. Column-wise missing value counts
missing_count = df.isnull().sum()

# 3. Percentage of missing values
missing_percent = (df.isnull().sum() / len(df)) * 100

# 4. Missing-value summary table
summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing %": missing_percent.round(2)
})
summary = summary[summary["Missing Count"] > 0].sort_values("Missing %", ascending=False)
print(summary)

# 5. Columns requiring treatment
print("\nColumns needing treatment:", list(summary.index))

```
